```python


from scgraph_data.world_highways_and_marnet import world_highways_and_marnet_geograph
from scgraph_data.world_railways import world_railways_geograph

try:
    import gurobipy as gp
    from gurobipy import GRB
except Exception:
    gp = None
    GRB = None


@contextlib.contextmanager
def _suppress_output():
    buf_out = io.StringIO()
    buf_err = io.StringIO()
    with contextlib.redirect_stdout(buf_out), contextlib.redirect_stderr(buf_err):
        yield

class UnifiedTSP:


    def __init__(self, mode="both"):
        assert mode in ("road", "rail", "both")
        self.mode = mode
        self.hwy = world_highways_and_marnet_geograph
        self.rail = world_railways_geograph


    def _pair_shortest(self, origin, dest):
        best_d = math.inf
        best_path = []
        best_mode = None

        # road/maritime
        if self.mode in ("road", "both"):
            with _suppress_output():
                try:
                    r = self.hwy.get_shortest_path(origin_node=origin, destination_node=dest, distance="km")
                    d = float(r.get("length", math.inf))
                    if d < best_d:
                        best_d = d
                        best_path = r.get("coordinate_path", []) or []
                        best_mode = "road/waterway"
                except Exception:
                    pass

        # rail
        if self.mode in ("rail", "both"):
            with _suppress_output():
                try:
                    r = self.rail.get_shortest_path(origin_node=origin, destination_node=dest, distance="km")
                    d = float(r.get("length", math.inf))
                    if d < best_d:
                        best_d = d
                        best_path = r.get("coordinate_path", []) or []
                        best_mode = "railway"
                except Exception:
                    pass

        if best_mode is None:
            return math.inf, [], None
        return best_d, best_path, best_mode


    def build_meta_graph(self, nodes, origin):
        def key(n):
            return (float(n["latitude"]), float(n["longitude"]))

        seen = set()
        coords = []
        seen.add(key(origin))
        coords.append(origin)
        for n in nodes:
            k = key(n)
            if k not in seen:
                seen.add(k)
                coords.append(n)

        for i, c in enumerate(coords, start=1):
            if not c.get("name"):
                c["name"] = f"Input {i}"

        id_to_node = {i: coords[i-1] for i in range(1, len(coords)+1)}
        node_ids = list(id_to_node.keys())

        dist = {u: {} for u in node_ids}
        path = {u: {} for u in node_ids}
        mode_used = {u: {} for u in node_ids}

        for u in node_ids:
            for v in node_ids:
                if u == v:
                    dist[u][v] = 0.0
                    path[u][v] = [id_to_node[u]]
                    mode_used[u][v] = None
                else:
                    d, p, m = self._pair_shortest(id_to_node[u], id_to_node[v])
                    dist[u][v] = d
                    path[u][v] = p or []
                    mode_used[u][v] = m

        return node_ids, id_to_node, dist, path, mode_used

    # ---------------------------
    # normalize segment coordinates to [lat, lon]
    # ---------------------------
    def _normalize_segment(self, seg):
        out = []
        for p in seg:
            if isinstance(p, dict):
                try:
                    out.append([float(p["latitude"]), float(p["longitude"])])
                except Exception:
                    continue
            elif isinstance(p, (list, tuple)) and len(p) >= 2:
                try:
                    out.append([float(p[0]), float(p[1])])
                except Exception:
                    continue
            else:
                continue
        return out

    # ---------------------------
    # Anti-meridian helpers (adjust longitudes to avoid giant jumps)
    # ---------------------------
    def adjustArcPath(self, path):
        # path: list of [lat, lon]
        for i in range(1, len(path)):
            lon = path[i][1]
            prev_lon = path[i-1][1]
            # shift by a multiple of 360 so that difference fits within [-180,180]
            path[i][1] = lon - (round((lon - prev_lon)/360) * 360)
        return path

    def modifyArcPathLong(self, pts, amt):
        return [[p[0], p[1] + amt] for p in pts]

    def getCleanArcPaths(self, path):
        # returns a list of candidate corrected paths without 180+ jumps
        base = self.adjustArcPath([p.copy() for p in path])
        candidates = [
            base,
            self.modifyArcPathLong([p.copy() for p in base], 360),
            self.modifyArcPathLong([p.copy() for p in base], -360),
            self.modifyArcPathLong([p.copy() for p in base], 720),
            self.modifyArcPathLong([p.copy() for p in base], -720)
        ]
        valid = []
        for c in candidates:
            if len(c) < 2:
                continue
            jumps = [abs(c[i][1] - c[i-1][1]) for i in range(1, len(c))]
            if max(jumps) < 180:
                valid.append(c)
        return valid if valid else [path]

    # ---------------------------
    # chain segments using SCGraph geometry and apply anti-meridian fix
    # ---------------------------
    def _chain_full_route(self, route_ids, path):
        full = []
        for a, b in zip(route_ids[:-1], route_ids[1:]):
            seg = path[a][b] or []
            seg_norm = self._normalize_segment(seg)
            if not seg_norm:
                continue
            if not full:
                full.extend(seg_norm)
            else:
                # avoid repeating join point duplicate
                if seg_norm[0][0] == full[-1][0] and seg_norm[0][1] == full[-1][1]:
                    full.extend(seg_norm[1:])
                else:
                    full.extend(seg_norm)
        # now fix anti-meridian wrap by splitting into cleaned candidate, pick first valid
        if not full:
            return []
        clean = self.getCleanArcPaths(full)
        # If multiple candidate paths, pick the one with smallest total lon-jumps (heuristic)
        def jump_metric(p):
            return sum(abs(p[i][1]-p[i-1][1]) for i in range(1, len(p)))
        best = min(clean, key=jump_metric)
        return best

    # ---------------------------
    # folium map builder (plots the actual geometry)
    # ---------------------------
    def _build_map(self, full_coords, visit_order_nodes):
        if not full_coords:
            center = [visit_order_nodes[0]["latitude"], visit_order_nodes[0]["longitude"]]
            m = folium.Map(location=center, zoom_start=5)
            for idx, n in enumerate(visit_order_nodes, start=1):
                folium.CircleMarker([n["latitude"], n["longitude"]], radius=6).add_to(m)
            return m

        # center on first coord
        m = folium.Map(location=[full_coords[0][0], full_coords[0][1]], zoom_start=5)

        # To preserve mode colors later: plot full polyline as red by default
        folium.PolyLine(full_coords, color="red", weight=3.5, opacity=0.9).add_to(m)

        # markers for visit_order_nodes
        for idx, n in enumerate(visit_order_nodes, start=1):
            lat = float(n["latitude"]); lon = float(n["longitude"])
            lab = f"<div style='background:white;padding:4px;border-radius:6px;font-weight:700'>{idx}. {n['name']} ({round(lat,4)},{round(lon,4)})</div>"
            folium.CircleMarker([lat, lon], radius=6, color="black", fill=True, fill_opacity=1).add_to(m)
            folium.Marker([lat, lon], icon=folium.DivIcon(html=lab)).add_to(m)
        return m

    # ---------------------------
    # greedy solver internal
    # ---------------------------
    def _greedy(self, node_ids, dist, start_id, tsp=True):
        route = [start_id]
        unvisited = set(node_ids) - {start_id}
        while unvisited:
            u = route[-1]
            # choose nearest finite distance, if all are inf choose the smallest placeholder (but will be detected)
            def keyfunc(v):
                dv = dist[u][v]
                return (math.isinf(dv), dv if not math.isinf(dv) else 1e12)
            nxt = min(unvisited, key=keyfunc)
            route.append(nxt)
            unvisited.remove(nxt)
        if tsp:
            route.append(start_id)
        return route

    # ---------------------------
    # ACO internal (uses LARGE for unreachable pairs but we will check route for inf)
    # ---------------------------
    def _aco(self, node_ids, dist, start_id, tsp=True, iterations=120, n_ants=None, alpha=1.0, beta=3.0, rho=0.1, q=1.0, seed=None):
        if seed is not None:
            np.random.seed(seed); random.seed(seed)
        n = len(node_ids)
        id_to_idx = {node_ids[i]: i for i in range(n)}
        idx_to_id = {i: node_ids[i] for i in range(n)}

        LARGE = 1e9
        D = np.zeros((n, n))
        for u in node_ids:
            for v in node_ids:
                D[id_to_idx[u], id_to_idx[v]] = (dist[u][v] if dist[u][v] != math.inf else LARGE)

        if n_ants is None:
            n_ants = max(10, n)

        pher = np.ones((n, n))
        heuristic = 1.0 / (D + 1e-12)

        best_route_idx = None
        best_len = float("inf")
        start_idx = id_to_idx[start_id]

        for _ in range(iterations):
            for _ in range(n_ants):
                route_idx = [start_idx]
                unv = set(range(n)) - {start_idx}
                while unv:
                    i = route_idx[-1]
                    cand = list(unv)
                    probs = np.array([(pher[i][j] ** alpha) * (heuristic[i][j] ** beta) for j in cand])
                    s = probs.sum()
                    if s <= 0:
                        probs = np.ones_like(probs) / len(probs)
                    else:
                        probs = probs / s
                    nxt = np.random.choice(cand, p=probs)
                    route_idx.append(nxt)
                    unv.remove(nxt)
                # length
                L = sum(D[route_idx[i]][route_idx[i+1]] for i in range(len(route_idx)-1))
                if tsp:
                    L += D[route_idx[-1]][route_idx[0]]
                if L < best_len:
                    best_len = L
                    best_route_idx = route_idx.copy()

                # update
                pher *= (1 - rho)
                deposit = q / (L + 1e-12)
                for a,b in zip(route_idx[:-1], route_idx[1:]):
                    pher[a][b] += deposit
                if tsp:
                    pher[route_idx[-1]][route_idx[0]] += deposit

        route_ids = [idx_to_id[i] for i in best_route_idx]
        if tsp:
            route_ids.append(route_ids[0])
        return route_ids

    # ---------------------------
    # Gurobi internal (we still substitute BIG in objective but will validate route after)
    # ---------------------------
    def _gurobi(self, node_ids, dist, start_id, time_limit=60, mip_gap=1e-4, verbose=False):
        if gp is None:
            raise RuntimeError("gurobipy not installed or license inactive.")

        ids = list(node_ids)
        N = len(ids)
        if N == 1:
            return [ids[0]]

        BIG = 1e9  # only for objective coefficients

        model = gp.Model("TSP")
        if not verbose:
            model.setParam("OutputFlag", 0)
        model.setParam("TimeLimit", float(time_limit))
        model.setParam("MIPGap", float(mip_gap))

        x = {}
        for i in ids:
            for j in ids:
                x[i,j] = model.addVar(vtype=GRB.BINARY, name=f"x_{i}_{j}")

        non_start = [i for i in ids if i != start_id]
        u = {}
        for i in non_start:
            u[i] = model.addVar(lb=1.0, ub=N-1.0, vtype=GRB.CONTINUOUS, name=f"u_{i}")

        model.update()

        obj = gp.quicksum((dist[i][j] if dist[i][j] != math.inf else BIG) * x[i,j] for i in ids for j in ids)
        model.setObjective(obj, GRB.MINIMIZE)

        for i in ids:
            model.addConstr(x[i,i] == 0)
        for i in ids:
            model.addConstr(gp.quicksum(x[i,j] for j in ids) == 1)
        for j in ids:
            model.addConstr(gp.quicksum(x[i,j] for i in ids) == 1)
        for i in non_start:
            for j in non_start:
                if i != j:
                    model.addConstr(u[i] - u[j] + N * x[i,j] <= N - 1)

        model.update()
        model.optimize()

        if model.Status not in (GRB.OPTIMAL, GRB.TIME_LIMIT):
            try:
                model.write("gurobi_debug_model.lp")
            except Exception:
                pass
            raise RuntimeError(f"Gurobi failed or infeasible (status={model.Status}). Debug LP saved if possible.")

        succ = {}
        for i in ids:
            for j in ids:
                try:
                    v = x[i,j].X
                except Exception:
                    v = 0.0
                if v > 0.5:
                    succ[i] = j

        # reconstruct route
        route = [start_id]
        cur = start_id
        visited = {start_id}
        for _ in range(N+5):
            nxt = succ.get(cur)
            if nxt is None:
                break
            route.append(nxt)
            cur = nxt
            if cur == start_id:
                break
            if cur in visited:
                try:
                    model.write("gurobi_debug_model.lp")
                except:
                    pass
                raise RuntimeError("Gurobi returned subtour (not full tour).")
            visited.add(cur)

        if len(route) != N+1:
            try:
                model.write("gurobi_debug_model.lp")
            except:
                pass
            raise RuntimeError("Gurobi produced incomplete route.")

        return route

    # ---------------------------
    # PUBLIC solve that enforces validation and anti-meridian fixes
    # ---------------------------
    def solve(self, nodes, origin, solver="gurobi", tsp=True, aco_params=None, gurobi_params=None):
        aco_params = aco_params or {}
        gurobi_params = gurobi_params or {}

        t_total_start = time.perf_counter()
        t_meta_start = time.perf_counter()
        node_ids, id_to_node, dist, path, mode_used = self.build_meta_graph(nodes, origin)
        t_meta_end = time.perf_counter()
        meta_graph_time = t_meta_end - t_meta_start

        if len(node_ids) == 1:
            total_time = time.perf_counter() - t_total_start
            return {
                "Mode_of_Solving": solver,
                "visit_order": [id_to_node[1]],
                "total_distance": 0.0,
                "segment_info": [],
                "full_route_coordinates": [],
                "folium_map": None,
                "meta_graph_time": round(meta_graph_time, 6),
                "solver_time": 0.0,
                "total_time": round(total_time, 6)
            }

        start_id = node_ids[0]

        t_solver_start = time.perf_counter()
        route_ids = None
        try:
            if solver.lower() == "gurobi":
                route_ids = self._gurobi(node_ids, dist, start_id,
                                         time_limit=gurobi_params.get("time_limit", 60),
                                         mip_gap=gurobi_params.get("mip_gap", 1e-4),
                                         verbose=gurobi_params.get("verbose", False))
            elif solver.lower() == "aco":
                route_ids = self._aco(node_ids, dist, start_id,
                                      tsp=tsp,
                                      iterations=aco_params.get("iterations", 120),
                                      n_ants=aco_params.get("n_ants"),
                                      alpha=aco_params.get("alpha", 1.0),
                                      beta=aco_params.get("beta", 3.0),
                                      rho=aco_params.get("rho", 0.1),
                                      q=aco_params.get("q", 1.0),
                                      seed=aco_params.get("seed", None))
            elif solver.lower() == "greedy":
                route_ids = self._greedy(node_ids, dist, start_id, tsp=tsp)
            else:
                raise ValueError("solver must be 'gurobi','aco' or 'greedy'")
        except Exception as e:
            t_solver_end_exc = time.perf_counter()
            solver_time_exc = t_solver_end_exc - t_solver_start
            total_time_exc = time.perf_counter() - t_total_start
            # raise informative error with timing info
            raise RuntimeError({
                "error": str(e),
                "Mode_of_Solving": solver,
                "meta_graph_time": round(meta_graph_time, 6),
                "solver_time": round(solver_time_exc, 6),
                "total_time": round(total_time_exc, 6)
            })

        t_solver_end = time.perf_counter()
        solver_time = t_solver_end - t_solver_start

        # Validate route: if any used edge has math.inf distance, report error
        bad_edges = []
        for a, b in zip(route_ids[:-1], route_ids[1:]):
            if math.isinf(dist[a][b]):
                bad_edges.append((a, b, id_to_node[a], id_to_node[b]))

        if bad_edges:
            # prepare message listing the problematic segments (names)
            msg_lines = []
            for (a,b,na,nb) in bad_edges:
                msg_lines.append(f"No path between '{na.get('name')}' and '{nb.get('name')}' for chosen mode")
            # raise with timings included (so caller can log meta/solver/total)
            raise RuntimeError({
                "error": " | ".join(msg_lines),
                "Mode_of_Solving": solver,
                "meta_graph_time": round(meta_graph_time, 6),
                "solver_time": round(solver_time, 6),
                "total_time": round(time.perf_counter() - t_total_start, 6),
                "bad_edges": [(id_to_node[a].get("name"), id_to_node[b].get("name")) for a,b,_,_ in bad_edges]
            })

        # Build visit order (exclude duplicate last element)
        visit_order_nodes = [id_to_node[i] for i in route_ids[:-1]]

        # Build segment info and total distance (safe sum, no BIG included)
        total_distance = 0.0
        segment_info = []
        for a,b in zip(route_ids[:-1], route_ids[1:]):
            na = id_to_node[a]; nb = id_to_node[b]
            d = dist[a][b]
            total_distance += d  # d is finite here because we validated
            segment_info.append({
                "from": f"{na['name']} ({na['latitude']}, {na['longitude']})",
                "to": f"{nb['name']} ({nb['latitude']}, {nb['longitude']})",
                "distance_km": d,
                "mode": mode_used[a][b]
            })

        # Build full route coordinates from stored segment paths and apply anti-meridian fix
        full_route_coords = self._chain_full_route(route_ids, path)

        # Build folium map plotting the real geometry
        folium_map = self._build_map(full_route_coords, visit_order_nodes)

        total_time = time.perf_counter() - t_total_start

        return {
            "Mode_of_Solving": solver,
            "visit_order": visit_order_nodes,
            "total_distance": float(total_distance),
            "segment_info": segment_info,
            "full_route_coordinates": full_route_coords,
            "folium_map": folium_map,
            "meta_graph_time": round(meta_graph_time, 6),
            "solver_time": round(solver_time, 6),
            "total_time": round(total_time, 6)
        }
```

```python
import math
import time
import random
import io
import contextlib
import numpy as np
import folium

from scgraph_data.world_highways_and_marnet import world_highways_and_marnet_geograph
from scgraph_data.world_railways import world_railways_geograph

try:
    import gurobipy as gp
    from gurobipy import GRB
except Exception:
    gp = None
    GRB = None


# silence SCGraph
@contextlib.contextmanager
def _suppress_output():
    buf_out = io.StringIO()
    buf_err = io.StringIO()
    with contextlib.redirect_stdout(buf_out), contextlib.redirect_stderr(buf_err):
        yield


class UnifiedTSPSolverGurobi:

    def __init__(self, mode="both"):
        assert mode in ("road", "rail", "both")
        self.mode = mode
        self.hwy = world_highways_and_marnet_geograph
        self.rail = world_railways_geograph

    # ----------------------------------------------------
    # shortest path between TWO points using SCGraph
    # ----------------------------------------------------
    def _pair_shortest(self, origin, dest):
        best_d = math.inf
        best_path = []
        best_mode = None

        # road / maritime
        if self.mode in ("road", "both"):
            with _suppress_output():
                try:
                    r = self.hwy.get_shortest_path(origin_node=origin,
                                                   destination_node=dest,
                                                   distance="km")
                    d = float(r.get("length", math.inf))
                    if d < best_d:
                        best_d = d
                        best_path = r.get("coordinate_path", []) or []
                        best_mode = "road/waterway"
                except Exception:
                    pass

        # rail
        if self.mode in ("rail", "both"):
            with _suppress_output():
                try:
                    r = self.rail.get_shortest_path(origin_node=origin,
                                                    destination_node=dest,
                                                    distance="km")
                    d = float(r.get("length", math.inf))
                    if d < best_d:
                        best_d = d
                        best_path = r.get("coordinate_path", []) or []
                        best_mode = "railway"
                except Exception:
                    pass

        if best_mode is None:
            return math.inf, [], None

        return best_d, best_path, best_mode

    # ----------------------------------------------------
    # Build meta graph: origin first, no duplicates
    # ----------------------------------------------------
    def build_meta_graph(self, nodes, origin):

        def key(n):
            return (float(n["latitude"]), float(n["longitude"]))

        coords = []
        seen = set()

        # Insert origin first
        seen.add(key(origin))
        coords.append(origin)

        # Insert remaining nodes, skipping duplicates
        for n in nodes:
            k = key(n)
            if k not in seen:
                seen.add(k)
                coords.append(n)

        # Add default names
        for i, c in enumerate(coords, start=1):
            if "name" not in c or not c["name"]:
                c["name"] = f"Input {i}"

        id_to_node = {i: coords[i-1] for i in range(1, len(coords)+1)}
        node_ids = list(id_to_node.keys())

        # matrices
        dist = {u: {} for u in node_ids}
        path = {u: {} for u in node_ids}
        mode_used = {u: {} for u in node_ids}

        for u in node_ids:
            for v in node_ids:
                if u == v:
                    dist[u][v] = 0.0
                    path[u][v] = [id_to_node[u]]
                    mode_used[u][v] = None
                else:
                    d, p, m = self._pair_shortest(id_to_node[u], id_to_node[v])
                    dist[u][v] = d
                    path[u][v] = p or []
                    mode_used[u][v] = m

        return node_ids, id_to_node, dist, path, mode_used

    # ----------------------------------------------------
    # Normalize geometry
    # ----------------------------------------------------
    def _normalize_segment(self, seg):
        out = []
        for p in seg:
            if isinstance(p, dict):
                try:
                    out.append([float(p["latitude"]), float(p["longitude"])])
                except:
                    pass
            elif isinstance(p, (list, tuple)) and len(p) >= 2:
                try:
                    out.append([float(p[0]), float(p[1])])
                except:
                    pass
        return out

    # -------------- Anti-meridian --------------
    def adjustArcPath(self, path):
        for i in range(1, len(path)):
            lon = path[i][1]
            prev = path[i-1][1]
            path[i][1] = lon - (round((lon - prev) / 360) * 360)
        return path

    def modifyArcPathLong(self, pts, amt):
        return [[p[0], p[1] + amt] for p in pts]

    def getCleanArcPaths(self, path):
        base = self.adjustArcPath([p[:] for p in path])
        cands = [
            base,
            self.modifyArcPathLong([p[:] for p in base], 360),
            self.modifyArcPathLong([p[:] for p in base], -360),
            self.modifyArcPathLong([p[:] for p in base], 720),
            self.modifyArcPathLong([p[:] for p in base], -720)
        ]

        good = []
        for c in cands:
            if len(c) < 2:
                continue
            jumps = [abs(c[i][1] - c[i-1][1]) for i in range(1, len(c))]
            if max(jumps) < 180:
                good.append(c)

        return good if good else [path]

    # ----------------------------------------------------
    # Chain full geometry path
    # ----------------------------------------------------
    def _chain_full_route(self, route_ids, path):
        full = []
        for a, b in zip(route_ids[:-1], route_ids[1:]):
            seg = self._normalize_segment(path[a][b])
            if not seg:
                continue
            if not full:
                full.extend(seg)
            else:
                if seg[0] == full[-1]:
                    full.extend(seg[1:])
                else:
                    full.extend(seg)

        if not full:
            return []

        cands = self.getCleanArcPaths(full)
        def metric(p):
            return sum(abs(p[i][1] - p[i-1][1]) for i in range(1, len(p)))

        return min(cands, key=metric)

    # ----------------------------------------------------
    # Map: markers + labels always shown
    # ----------------------------------------------------
    def _build_map(self, full_coords, visit_order_nodes):
        raw_path = []
        for p in full_coords:
            if isinstance(p, dict):
                raw_path.append([float(p["latitude"]), float(p["longitude"])])
            else:
                raw_path.append([float(p[0]), float(p[1])])

        clean_paths = self.getCleanArcPaths(raw_path)

        # map center on origin
        lat0 = float(visit_order_nodes[0]["latitude"])
        lon0 = float(visit_order_nodes[0]["longitude"])
        m = folium.Map(location=[lat0, lon0], zoom_start=6)

        # draw polylines
        for cp in clean_paths:
            folium.PolyLine(cp, color="red", weight=4).add_to(m)

        # draw node markers
        for idx, n in enumerate(visit_order_nodes, start=1):
            lat = float(n["latitude"]); lon = float(n["longitude"])

            folium.CircleMarker(
                [lat, lon], radius=6, color="black", fill=True).add_to(m)

            label = (
                f"<div style='background:white;padding:3px;border-radius:4px;"
                f"font-size:12px;font-weight:bold;'>"
                f"{idx}. {n['name']} ({round(lat,4)}, {round(lon,4)})</div>"
            )

            folium.Marker(
                [lat, lon], icon=folium.DivIcon(html=label)
            ).add_to(m)

        return m

    # ----------------------------------------------------
    # Greedy solver (open/closed supported)
    # ----------------------------------------------------
    def _greedy(self, node_ids, dist, start_id, tsp=True):
        route = [start_id]
        unv = set(node_ids) - {start_id}

        while unv:
            u = route[-1]
            nxt = min(unv, key=lambda v: (math.isinf(dist[u][v]), dist[u][v]))
            route.append(nxt)
            unv.remove(nxt)

        if tsp:
            route.append(start_id)

        return route

    # ----------------------------------------------------
    # ACO solver (already supports tsp flag)
    # ----------------------------------------------------
    def _aco(self, node_ids, dist, start_id,
             tsp=True, iterations=120, n_ants=None,
             alpha=1, beta=3, rho=0.1, q=1, seed=None):

        if seed is not None:
            np.random.seed(seed); random.seed(seed)

        n = len(node_ids)
        id2i = {node_ids[i]: i for i in range(n)}
        i2id = {i: node_ids[i] for i in range(n)}

        LARGE = 1e9
        D = np.zeros((n, n))
        for u in node_ids:
            for v in node_ids:
                D[id2i[u], id2i[v]] = dist[u][v] if dist[u][v] != math.inf else LARGE

        pher = np.ones((n, n))
        heuristic = 1.0 / (D + 1e-12)

        if n_ants is None:
            n_ants = max(10, n)

        best_route = None
        best_len = float("inf")
        start_idx = id2i[start_id]

        for _ in range(iterations):
            for _a in range(n_ants):

                route = [start_idx]
                unv = set(range(n)) - {start_idx}

                while unv:
                    i = route[-1]
                    cand = list(unv)

                    probs = np.array([(pher[i][j] ** alpha) * (heuristic[i][j] ** beta)
                                      for j in cand])
                    s = probs.sum()
                    if s <= 0:
                        probs = np.ones_like(probs)/len(probs)
                    else:
                        probs = probs/s

                    nxt = np.random.choice(cand, p=probs)
                    route.append(nxt)
                    unv.remove(nxt)

                L = sum(D[route[i]][route[i+1]] for i in range(len(route)-1))
                if tsp:
                    L += D[route[-1]][route[0]]

                if L < best_len:
                    best_len = L
                    best_route = route.copy()

                pher *= (1 - rho)
                d = q/(L + 1e-12)
                for a, b in zip(route[:-1], route[1:]):
                    pher[a][b] += d
                if tsp:
                    pher[route[-1]][route[0]] += d

        out = [i2id[i] for i in best_route]
        if tsp:
            out.append(out[0])

        return out

    # ----------------------------------------------------
    # Gurobi solver (fixed to support open & closed TSP)
    # ----------------------------------------------------
    def _gurobi(self, node_ids, dist, start_id,
                time_limit=60, mip_gap=1e-4, verbose=False, tsp=True):

        if gp is None:
            raise RuntimeError("Install Gurobi + license.")

        ids = node_ids[:]
        N = len(ids)
        BIG = 1e9

        m = gp.Model("TSP")
        if not verbose:
            m.setParam("OutputFlag", 0)
        m.setParam("TimeLimit", float(time_limit))
        m.setParam("MIPGap", float(mip_gap))

        x = {}
        for i in ids:
            for j in ids:
                x[i, j] = m.addVar(vtype=GRB.BINARY)

        # MTZ u variables for subtour elimination (non-start nodes only)
        non = [i for i in ids if i != start_id]
        u = {}
        for i in non:
            u[i] = m.addVar(lb=1.0, ub=N-1.0, vtype=GRB.CONTINUOUS)

        m.update()

        m.setObjective(gp.quicksum(
            (dist[i][j] if dist[i][j] != math.inf else BIG) * x[i, j]
            for i in ids for j in ids
        ), GRB.MINIMIZE)

        # no self loops
        for i in ids:
            m.addConstr(x[i, i] == 0)

        # degree constraints
        if tsp:
            # CLOSED: each node must have exactly 1 outgoing AND 1 incoming
            for i in ids:
                m.addConstr(gp.quicksum(x[i, j] for j in ids) == 1)
                m.addConstr(gp.quicksum(x[j, i] for j in ids) == 1)
        else:
            # OPEN TSP (start fixed, end free)
            # start: out=1, in=0
            m.addConstr(gp.quicksum(x[start_id, j] for j in ids) == 1)
            m.addConstr(gp.quicksum(x[j, start_id] for j in ids) == 0)

            # others: in<=1, out<=1
            for i in ids:
                if i != start_id:
                    m.addConstr(gp.quicksum(x[i, j] for j in ids) <= 1)
                    m.addConstr(gp.quicksum(x[j, i] for j in ids) <= 1)

            # force exactly N-1 edges selected → Hamiltonian path
            m.addConstr(gp.quicksum(x[i, j] for i in ids for j in ids) == N - 1)

        # MTZ subtour elimination
        for i in non:
            for j in non:
                if i != j:
                    m.addConstr(u[i] - u[j] + N * x[i, j] <= N - 1)

        m.optimize()

        if m.Status not in (GRB.OPTIMAL, GRB.TIME_LIMIT):
            raise RuntimeError("Gurobi failed.")

        # successor mapping
        succ = {}
        for i in ids:
            for j in ids:
                if x[i, j].X > 0.5:
                    succ[i] = j

        # reconstruct path
        route = [start_id]
        curr = start_id
        visited = {start_id}

        for _ in range(N + 5):
            nxt = succ.get(curr)
            if nxt is None:
                break
            route.append(nxt)
            if tsp and nxt == start_id:
                break
            if nxt in visited:
                raise RuntimeError("Subtour found.")
            visited.add(nxt)
            curr = nxt

        # validate route length
        if tsp:
            if len(route) != N + 1 or route[-1] != start_id:
                raise RuntimeError("Incomplete closed tour.")
        else:
            if len(route) != N:
                raise RuntimeError("Incomplete open path.")

        return route

    # ----------------------------------------------------
    # PUBLIC SOLVE()
    # ----------------------------------------------------
    def solve(self, nodes, origin, solver="gurobi", tsp=True,
              aco_params=None, gurobi_params=None):

        aco_params = aco_params or {}
        gurobi_params = gurobi_params or {}

        t_total = time.perf_counter()

        # meta graph
        t_meta = time.perf_counter()
        node_ids, id_to_node, dist, path, mode_used = self.build_meta_graph(nodes, origin)
        meta_graph_time = time.perf_counter() - t_meta

        start_id = node_ids[0]

        # solver
        t_solver = time.perf_counter()

        if solver.lower() == "gurobi":
            route_ids = self._gurobi(
                node_ids, dist, start_id,
                time_limit=gurobi_params.get("time_limit", 60),
                mip_gap=gurobi_params.get("mip_gap", 1e-4),
                verbose=gurobi_params.get("verbose", False),
                tsp=tsp
            )

        elif solver.lower() == "aco":
            route_ids = self._aco(
                node_ids, dist, start_id,
                tsp=tsp,
                iterations=aco_params.get("iterations", 120),
                n_ants=aco_params.get("n_ants"),
                alpha=aco_params.get("alpha", 1.0),
                beta=aco_params.get("beta", 3.0),
                rho=aco_params.get("rho", 0.1),
                q=aco_params.get("q", 1.0),
                seed=aco_params.get("seed")
            )

        elif solver.lower() == "greedy":
            route_ids = self._greedy(node_ids, dist, start_id, tsp=tsp)

        else:
            raise ValueError("Unknown solver.")

        solver_time = time.perf_counter() - t_solver

        # validate edges
        for a, b in zip(route_ids[:-1], route_ids[1:]):
            if math.isinf(dist[a][b]):
                raise RuntimeError(f"UNREACHABLE EDGE between {id_to_node[a]['name']} and {id_to_node[b]['name']}")

        # visit order nodes
        if tsp:
            visit_order_nodes = [id_to_node[i] for i in route_ids[:-1]]
        else:
            visit_order_nodes = [id_to_node[i] for i in route_ids]

        # total distance
        total_distance = sum(dist[a][b] for a, b in zip(route_ids[:-1], route_ids[1:]))

        # segment info
        segment_info = []
        for a, b in zip(route_ids[:-1], route_ids[1:]):
            na, nb = id_to_node[a], id_to_node[b]
            segment_info.append({
                "from": f"{na['name']} ({na['latitude']}, {na['longitude']})",
                "to": f"{nb['name']} ({nb['latitude']}, {nb['longitude']})",
                "distance_km": dist[a][b],
                "mode": mode_used[a][b]
            })

        # geometry path
        full_route_coords = self._chain_full_route(route_ids, path)

        folium_map = self._build_map(full_route_coords, visit_order_nodes)

        total_time = time.perf_counter() - t_total

        return {
            "Mode_of_Solving": solver,
            "visit_order": visit_order_nodes,
            "total_distance": float(total_distance),
            "segment_info": segment_info,
            "full_route_coordinates": full_route_coords,
            "folium_map": folium_map,
            "meta_graph_time": round(meta_graph_time, 6),
            "solver_time": round(solver_time, 6),
            "total_time": round(total_time, 6)
        }
```
