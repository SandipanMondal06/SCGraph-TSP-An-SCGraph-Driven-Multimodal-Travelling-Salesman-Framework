# GridGraph - using SCGraph

```python
from scgraph import GridGraph
x_size = 20
y_size = 20
blocks = [(10 , i ) for i in range (5 , y_size ) ]
gridGraph = GridGraph (
x_size = x_size ,
y_size = y_size ,
blocks = blocks ,
add_exterior_walls = True ,
)
output = gridGraph . get_shortest_path (
origin_node ={"x": 2 , "y": 10} ,
destination_node ={"x": 18 , "y": 10} ,
)
print(output)
```

# visualize_gridgraph(x_size, y_size, blocks, origin, destination, output)

This visualisation tool I have developed using AI

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

def visualize_gridgraph(x_size, y_size, blocks, origin, destination, output,
                        show_coords=False, invert_y=True, save_path=None):

    coordinate_path = output.get("coordinate_path", [])
    length = output.get("length", None)

    fig, ax = plt.subplots(figsize=(7, 7))

    # Draw grid
    for x in range(x_size + 1):
        ax.plot([x, x], [0, y_size], linewidth=0.5, color='gray', alpha=0.5)
    for y in range(y_size + 1):
        ax.plot([0, x_size], [y, y], linewidth=0.5, color='gray', alpha=0.5)

    # Draw blocked cells
    for bx, by in blocks:
        ax.add_patch(Rectangle((bx, by), 1, 1, color='orange', alpha=0.9))

    # Plot the path
    if coordinate_path:
        # Extract x and y coordinates from the list of dictionaries
        xs = [p['x'] + 0.5 for p in coordinate_path]
        ys = [p['y'] + 0.5 for p in coordinate_path]
        ax.plot(xs, ys, '-o', linewidth=2.5, markersize=5, color='blue', label='Path')

        if show_coords:
            for (px, py) in coordinate_path:
                ax.text(px + 0.5, py + 0.5, f'({px},{py})', fontsize=6,
                        ha='center', va='center')

    # Mark origin and destination
    ax.scatter(origin["x"] + 0.5, origin["y"] + 0.5, s=100, marker='D', color='green', label='Origin')
    ax.scatter(destination["x"] + 0.5, destination["y"] + 0.5, s=100, marker='X', color='red', label='Destination')

    # Axis settings
    ax.set_xlim(0, x_size)
    ax.set_ylim(0, y_size)
    ax.set_aspect('equal')
    if invert_y:
        ax.invert_yaxis()

    ax.set_xticks(range(x_size + 1))
    ax.set_yticks(range(y_size + 1))
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_title(f'Shortest Path (length ≈ {length:.4f})')
    ax.legend(loc='upper right')

    plt.show()
```

```python
visualize_gridgraph(x_size,y_size,blocks, {"x": 2 , "y": 10},{"x": 18 , "y": 10},output)
```
